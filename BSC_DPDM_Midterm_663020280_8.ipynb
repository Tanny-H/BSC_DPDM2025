##กรองเฉพาะสถานีใน Stations in the model_HII.csv

```python
stations_df = pd.read_csv(
    "/content/drive/MyDrive/bscdpdmmid25/input/Stations in the model_HII.csv"
)

stations_df.columns
```

```python
stations_df = stations_df.rename(columns={"Node": "station"})
stations_df["station"] = stations_df["station"].astype(str).str.strip()
```

กรองข้อมูลฝน

```python
before = rain_df["station"].nunique()

rain_df = rain_df[rain_df["station"].isin(stations_df["station"])]

after = rain_df["station"].nunique()

print("Stations before filtering:", before)
print("Stations after filtering :", after)
```

```python
full_dates = pd.date_range(
    start="2025-07-01",
    end="2025-12-31",
    freq="D"
)

expected_days = len(full_dates)  # ≈ 184
expected_days
```

```python
actual_days = (
    rain_df
    .groupby("station")["measure_datetime"]
    .nunique()
    .rename("actual_days")
)
```

##ระบุข้อมูลสูญหายของแต่ละสถานี

```python
missing_summary = (
    rain_df
    .groupby("station")["rain"]
    .agg(
        total_records="size",
        missing_records=lambda x: x.isna().sum()
    )
)

missing_summary["missing_ratio"] = (
    missing_summary["missing_records"] /
    missing_summary["total_records"]
)

missing_summary.describe()
```

```python
missing_summary = pd.DataFrame({
    "actual_days": actual_days
})

missing_summary["expected_days"] = expected_days
missing_summary["missing_records"] = (
    missing_summary["expected_days"] -
    missing_summary["actual_days"]
)

missing_summary["missing_ratio"] = (
    missing_summary["missing_records"] /
    missing_summary["expected_days"]
)

missing_summary.describe()
```

```python
missing_summary.sort_values("missing_ratio", ascending=False).head(10)
```

##คัดสถานีที่ missing ≥ 20% ออก

```python
bad_stations = missing_summary[
    missing_summary["missing_ratio"] >= 0.20
].index

print("Stations removed:", len(bad_stations))
```

```python
rain_df = rain_df[~rain_df["station"].isin(bad_stations)]

print("Stations remaining:", rain_df["station"].nunique())
```

```python
rain_df.groupby("station")["measure_datetime"].nunique().describe()
```

##เติมข้อมูลด้วย Median ตาม Day of Year

```python
median_doy = (
    rain_df
    .groupby(["station", "doy"])["rain"]
    .median()
)
```

เติม missing

```python
def fill_median(row):
    if pd.isna(row["rain"]):
        return median_doy.get((row["station"], row["doy"]), np.nan)
    return row["rain"]

rain_df["rain"] = rain_df.apply(fill_median, axis=1)
```

ตรวจสอบ

```python
rain_df["rain"].isna().sum()
```

ใช้ median ของสถานีนั้นแทน

```python
station_median = rain_df.groupby("station")["rain"].median()
```

```python
rain_df.loc[:, "rain"] = rain_df.apply(
    lambda row: (
        station_median[row["station"]]
        if pd.isna(row["rain"])
        else row["rain"]
    ),
    axis=1
)
```

ตรวบสอบ

```python
rain_df["rain"].isna().sum()
```

##อภิปราย: เหตุผลที่เลือกใช้ Median แทน Mean ในการจัดการ Missing Data
ในการวิเคราะห์ข้อมูลปริมาณน้ำฝน พบว่าข้อมูลมีลักษณะการกระจายที่ ไม่เป็นแบบปกติ (non-normal distribution) กล่าวคือ ในหลายสถานีจะมีค่าปริมาณน้ำฝนเป็นศูนย์หรือมีค่าน้อยเป็นส่วนใหญ่ แต่บางวันอาจเกิดฝนตกหนักมาก ส่งผลให้มีค่าที่สูงผิดปกติ (outliers)

หากใช้ ค่าเฉลี่ย (Mean) ในการแทนค่าที่หายไป ค่าฝนที่สูงผิดปกติเหล่านี้จะดึงค่าเฉลี่ยให้สูงขึ้น ทำให้ค่าที่นำมาแทน ไม่สะท้อนพฤติกรรมปกติของข้อมูลจริง และอาจทำให้ผลการวิเคราะห์คลาดเคลื่อน

ค่ามัธยฐาน (Median) เป็นค่ากลางที่ไม่อ่อนไหวต่อ outliers จึงสามารถแทนค่าที่หายไปได้อย่างเหมาะสมกว่า โดยเฉพาะในข้อมูลปริมาณน้ำฝนซึ่งมีลักษณะเป็นข้อมูลเชิงสิ่งแวดล้อมที่มีความแปรปรวนสูง
